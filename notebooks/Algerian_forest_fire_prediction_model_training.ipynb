```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset_cleaned.csv')
df.head()
```

```python
df['DC'] = df['DC'].fillna(df['DC'].mean())
```

```python
df['FWI'] = df['FWI'].fillna(df['FWI'].mean())
```

```python
df = df.drop(['day','month','year'],axis=1)
```

```python
df.head()
```

```python
#devide dataset independent and dependent feature
X = df.drop('FWI',axis=1)
y = df[['FWI']]
```

```python
##spliting to test and train data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
##standardization of the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_standard = scaler.fit_transform(X_train)
X_test_standard = scaler.transform(X_test)
X_train_standard
```

```python
sns.boxplot(X_train_standard)
```

```python
X_train.corr()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(X_train.corr(),annot=True)
```

```python
#removing the high correlated features as they can cause the overfittig
def remove_high_corr(dataset, threshold):
    high_corr_features = set()
    correlation = dataset.corr()
    for i in range(len(correlation.columns)):
        for j in range (i):
            if abs(correlation.iloc[i,j]) > threshold:
                high_corr = correlation.columns[i]
                high_corr_features.add(high_corr)
    return high_corr_features
```

```python
high_corr = remove_high_corr(X_train,0.98)
high_corr
```

```python
df.drop(high_corr,axis=1,inplace=True)
```

```python
df.head()
```

```python
X_train_standard
```

```python
##train the model using linear regression.
from sklearn.linear_model import LinearRegression
regression_linear = LinearRegression(n_jobs=-1)
regression_linear = regression_linear.fit(X_train_standard,y_train)
regression_linear
```

```python
print(regression_linear.coef_)
print(regression_linear.intercept_)
```

```python
y_pred_linear = regression_linear.predict(X_test_standard)
y_pred_linear
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
mae = mean_absolute_error(y_test,y_pred_linear)
r2_score = r2_score(y_test, y_pred_linear)
print(f"linear reggression mae: {mae}")
print(f"linear reggression r2_score: {r2_score}")   
```

```python
plt.scatter(y_test,y_pred_linear)
```

```python
#train data ki predict chesi error or accuracy find cheyadam
from sklearn.metrics import mean_absolute_error,r2_score
y_train_pred = regression_linear.predict(X_train_standard)
mae_train = mean_absolute_error(y_train,y_train_pred)
r2_score_train = r2_score(y_train,y_train_pred)
print(f"linear reggression mae_train: {mae_train}")
print(f"linear reggression r2_score_train: {r2_score_train}")
```

```python
## test data ki 98% accuracy isthundi ante overfitting ani ardam so apply ridge and lasso and elastic to decrease
from sklearn.linear_model import Ridge
ridge_regression = Ridge()
ridge = ridge_regression.fit(X_train_standard,y_train)
print(f"Ridge reggression cooeff: {ridge.coef_}")
print(f"Ride reggression intercept: {ridge.intercept_}")
```
